## 1. Question

# Research Question

## Can historical content freshness and search-performance signals help prioritize which pages should be reviewed first, while improving the precision of a transparent rule-based baseline on unseen clients?

### Decision supported

The decision supported by this work is:

> Given limited editorial or content-review capacity, which pages should be reviewed first?

Content teams cannot review every page at the same time. A prioritization system can help direct human attention toward pages showing combinations of historical freshness and search-performance signals associated with declining performance.

This project compares two approaches:

1. A transparent rule-based baseline using content freshness, search visibility, click-through rate, and search position.
2. A Random Forest model trained on historical features and evaluated on unseen clients.

The output is intended as a decision-support tool rather than an automated content-refresh system. A high-ranked page is a candidate for human review, not proof that the page requires changes.

### Primary evaluation question

Does the model improve the precision of the highest-priority review queue compared with the Week-4 transparent rule baseline when both are evaluated on the same client-grouped holdout?

### Why Precision@K matters

The practical workflow has limited review capacity. If a team can review only the top 20 or top 50 recommended pages, the quality of those ranked recommendations matters more than making an unrestricted prediction across every page.

For that reason, Precision@20 and Precision@50 are primary decision-support metrics, while accuracy, precision, recall, and F1 provide additional context about overall classification behavior.

## 3. Methodology

*Assumptions, features, label definition, baseline, validation design, leakage checks.*

## Study design

This work evaluates whether historical content and search-performance signals can improve the prioritization of pages for human review.

The comparison uses two approaches evaluated on the same client-grouped holdout:

1. A transparent rule-based baseline developed in Week 4.
2. A Random Forest classifier developed and evaluated in the Week 5 and Week 6 workflows.

The primary objective is not to automate content changes. The objective is to produce a ranked queue that helps a human team decide which pages to review first.

---

## Target definition

The target variable is:

`is_declining_label`

In the project workflow, this label represents whether the observed trend direction is:

`trend_direction == "down"`

The target is used for supervised evaluation but is not included as a predictive feature.

The related fields:

- `is_declining_label`
- `trend_direction`
- `trend_pct`

are excluded from the feature set because they directly define, derive, or reveal information closely connected to the target.

---

## Model features

The final Random Forest uses six historical features:

- `content_age_days`
- `days_since_last_update`
- `impressions_90d`
- `avg_position`
- `ctr`
- `word_count`

These features represent:

- content age,
- update recency,
- historical search visibility,
- average search position,
- click-through rate,
- and content length.

Identifiers such as `client_id` and `content_id` are not used as predictive features.

---

## Random Forest model

A Random Forest classifier is used as the machine-learning approach.

The model was selected as a practical nonlinear baseline capable of learning interactions between historical freshness and search-performance signals without requiring the prioritization rule to be manually specified.

The results in this paper should therefore be interpreted as measured performance for this model configuration on the provided dataset and validation design. They do not establish that Random Forest is universally the best model for content prioritization.

---

## Transparent rule baseline

The Week-4 baseline uses simple, explainable conditions based on historical signals.

A page receives positive signals when it is:

- stale, based on a long time since its last update,
- receiving meaningful search impressions,
- showing low click-through rate,
- or ranking outside the first page of search results.

The conditions are combined into a transparent baseline score rather than learned weights.

The baseline produces reason codes such as:

- `STALE_VISIBLE`
- `LOW_CTR`
- `LOW_VISIBILITY`
- `HIGH_PRIORITY_REVIEW`

This provides a meaningful comparison because the model is evaluated against a practical rule-based approach rather than against no baseline.

---

## Ranking and evaluation metrics

Both approaches are evaluated as ranked prioritization systems.

The primary metrics are:

- **Precision@20** — the proportion of truly declining pages among the top 20 ranked recommendations.
- **Precision@50** — the proportion of truly declining pages among the top 50 ranked recommendations.

These metrics reflect a realistic decision constraint: a human team may only have capacity to review a limited number of pages.

Additional classification metrics are reported for context:

- Accuracy
- Precision
- Recall
- F1 score

The model and baseline are evaluated on the same test rows and against the same labels.

---

## Validation design

Two evaluation perspectives were examined.

### Random-row evaluation

An earlier random-row split provides an initial estimate of performance when observations are randomly divided between training and testing.

However, this design can be optimistic when multiple observations from the same client appear in both splits.

### Client-grouped evaluation

The primary validation design uses a client-grouped holdout.

The dataset contains 32 anonymized clients. The grouped split contains:

- **25 training clients**
- **7 test clients**
- **0 client overlap**

This design tests whether patterns learned from the training clients can generalize to clients that were not included in training.

The client-grouped holdout is treated as the primary evaluation for the paper because it provides a stronger separation between training and evaluation entities than a random row split.

---

## Before-and-after validation comparison

The validation audit compared random-row evaluation with client-grouped evaluation.

Under the random-row split, the Random Forest achieved:

- Precision@20: 0.85
- Precision@50: 0.88
- Accuracy: 0.6835
- F1: 0.7192

Under the client-grouped holdout, performance was:

- Precision@20: 0.85
- Precision@50: 0.70
- Accuracy: 0.5629
- F1: 0.5928

The grouped evaluation therefore produced lower values for several broader metrics while Precision@20 remained unchanged.

This result is reported directly rather than treated as an error. It suggests that random-row evaluation can provide a more optimistic estimate of broader performance than evaluation on unseen clients.

---

## Leakage checks

A feature-level leakage audit was performed before the final grouped evaluation.

The final predictive feature set excludes:

- the target itself,
- fields used to define the target,
- target-derived trend information,
- client identifiers,
- and content identifiers.

A deliberate leakage diagnostic was also used to demonstrate the importance of these exclusions.

When target-derived trend information was introduced into the feature set, evaluation performance became artificially perfect at Precision@50. This behavior was treated as evidence of leakage rather than model quality.

The clean grouped model does not use those target-derived fields.

A column-level leakage audit does not prove complete temporal safety. The available dataset and feature semantics do not support a stronger claim that every possible production-time ordering or future-data pathway has been eliminated.

For this reason, the results are framed as an observational validation of the provided dataset and split design.

---

## Reproducibility assumptions

The same grouped holdout is used for the baseline and Random Forest comparison.

The primary methodological assumption is that the available historical features are appropriate for the decision-support setting represented by the dataset.

The results should be interpreted as measured associations and ranking performance on this dataset. They do not establish that changing a page will cause its search performance to improve, and they should not be interpreted as causal evidence.

## 5. Limitations

*What this work cannot claim.*


## This is an observational study

The results in this paper are based on patterns observed in the provided anonymized dataset.

The model identifies associations between historical content/search-performance signals and the declining-content label. The results do not establish that any individual feature causes content performance to decline.

For example, a page that has not been updated recently may appear frequently among declining pages, but this does not prove that updating the page will improve its performance.

---

## The label is not a complete measure of content quality

The supervised target is derived from:

`trend_direction == "down"`

A declining trend can have multiple possible explanations, including changes in search competition, seasonality, search-result layout, indexing, changes in user intent, or other factors not represented in the feature set.

Therefore, a positive model prediction should be interpreted as a recommendation for review, not as proof that the content itself is poor or requires a refresh.

---

## Generalization is evaluated on a limited number of clients

The primary client-grouped holdout contains:

- 25 training clients
- 7 test clients
- 0 client overlap

This provides a stronger evaluation than random-row splitting for the available dataset, but the results are still based on a limited number of held-out client groups.

Performance may differ for clients, content types, industries, traffic levels, or search environments that are not well represented in this dataset.

---

## Dataset-specific performance

The reported metrics are measured on the provided anonymized FlyRank ML Internship dataset.

The result that Precision@20 increased from 0.55 for the transparent baseline to 0.85 for the Random Forest should not be interpreted as a guaranteed improvement in every future production environment.

Model performance can change when the data distribution, client population, search behavior, or underlying search systems change.

---

## Historical metrics may not represent all production-time information

The feature set includes historical signals such as trailing 90-day impressions and observed search-performance measurements.

The project performs column-level leakage checks and excludes target-derived fields, but a column-level audit cannot prove complete temporal safety for every possible production deployment scenario.

A production implementation would require explicit feature timestamps, prediction timestamps, and a point-in-time data pipeline to guarantee that every feature available to the model existed before the prediction decision.

---

## Precision and recall involve a trade-off

The Random Forest improved the precision-oriented ranking metrics and slightly improved F1 relative to the rule baseline, but recall was lower:

- Baseline recall: 0.6932
- Random Forest recall: 0.6227

This means the model's prioritization strategy may miss some pages labeled as declining.

The model is therefore most appropriate for prioritizing a limited review queue, rather than for claiming to identify every declining page.

---

## The evaluation does not measure business impact

This project evaluates predictive and ranking performance.

It does not measure whether reviewing or refreshing the recommended pages produces:

- increased traffic,
- increased clicks,
- improved rankings,
- or other downstream business outcomes.

A production experiment would require intervention tracking and post-action measurement to evaluate whether recommendations lead to measurable improvements.

---

## The model does not replace human judgment

The output is designed for decision support.

A high-ranked page should be reviewed alongside context that may not be available in the model, such as:

- seasonality,

The recommended workflow is therefore:

> **Model ranking → human review → decision → action → post-action measurement**

## 6. Ranked recommendations

*The action playbook output — the paper's recommendations section.*


## Decision-support workflow

The purpose of the model is to prioritize content for human review.

The recommended workflow is:

> **Rank → Explain → Human review → Decide → Act → Measure**

The model does not automatically refresh, rewrite, publish, or delete content. A high-ranked page is a candidate for earlier inspection, not proof that a particular action will improve performance.

Recommendations are ranked using the model's decline probability and the observable signals associated with each page.

---

## Priority 1 — Review multi-signal candidates first

The highest-priority pages are candidates where multiple observed signals point in the same direction.

Typical reason codes may include combinations of:

- `AGE`
- `STALE`
- `VISIBILITY`
- `CTR`
- `PERFORMANCE`

These pages should receive a comprehensive human review because several observable characteristics support their priority.

**Recommended action:**  
Perform a full review of search intent, content accuracy, freshness, competitive context, technical factors, recent performance, and business value.

**Why this comes first:**  
Multiple signals provide more context for prioritization than a single observed characteristic.

**Important limitation:**  
Multiple signals do not prove that the content itself caused the observed decline.

---

## Priority 2 — Review freshness and update candidates

Pages with relatively high content age or a long time since their last update may receive the reason codes:

- `AGE`
- `STALE`

**Recommended action:**  
Review factual accuracy, topic freshness, changes in user needs, and opportunities to update the content.

**What to verify before acting:**

- Does the topic change frequently?
- Are important facts still accurate?
- Has the page already been updated recently?
- Is the content still aligned with current search intent?
- Is the observed performance pattern persistent?

**Honest interpretation:**  
Older or less recently updated content that also receives a high model-based decline probability deserves earlier human inspection. The observed relationship does not prove that aging caused the decline or that refreshing the page will improve performance.

---

## Priority 3 — Review visibility candidates

Pages with weaker observed search position may receive the reason code:

- `VISIBILITY`

**Recommended action:**  
Review search intent, page structure, internal linking, content coverage, and competitive context.

**What to verify before acting:**

- Has search intent changed?
- Have competing search results changed?
- Is the page appropriately structured for the topic?
- Could technical SEO or indexing explain the weaker position?
- Is the signal persistent rather than temporary?

A weaker position should trigger investigation rather than an automatic content refresh.

---

## Priority 4 — Review CTR candidates

Pages with relatively low click-through rate may receive the reason code:

- `CTR`

**Recommended action:**  
Review title and meta presentation, search-intent alignment, and how the page appears relative to competing search results.

**What to verify before acting:**

- Are titles and descriptions aligned with likely user intent?
- Have search-result layouts changed?
- Are rich results or other SERP features affecting clicks?
- Is low CTR persistent across the available observation window?

Low CTR alone does not prove that the underlying content requires rewriting.

---

## Priority 5 — Investigate performance candidates

Pages with relatively weak recent performance may receive the reason code:

- `PERFORMANCE`

**Recommended action:**  
Investigate recent performance and demand context before allocating content-refresh effort.

**What to verify before acting:**

- Is demand changing because of seasonality?
- Has search interest changed?
- Is the decline temporary?
- Are there external measurement or technical explanations?
- Does the page have sufficient business value to justify review effort?

The model ranking should help determine where to investigate first, not replace contextual analysis.

---

## Priority 6 — Monitor lower-priority candidates

Pages with lower model-based priority should generally be monitored rather than immediately consuming review capacity.

**Recommended action:**  
Track future performance and reconsider the page if its model ranking or observed signals change materially.

This allows limited human-review capacity to focus first on higher-ranked candidates.

---

## Reason codes make the queue interpretable

The action queue uses human-readable reason codes to explain the observed signals behind a recommendation:

| Reason code | Interpretation |
|---|---|
| `AGE` | Content age is relatively high in the modeling population |
| `STALE` | Time since the last update is relatively high |
| `VISIBILITY` | Average search position is relatively weak among pages with usable position data |
| `CTR` | Click-through rate is relatively low |
| `PERFORMANCE` | Recent impressions are relatively low |
| `MULTI` | Multiple signals point in the same direction |

Reason codes describe observed characteristics. They should not be interpreted as causal explanations.

---

## Required human-review gate

Before approving a content action, a reviewer should check:

1. **Search intent** — Does the page still answer the intended user need?
2. **Accuracy** — Are important claims and facts still correct?
3. **Freshness** — Has the underlying topic changed?
4. **Competitive context** — Have competing results changed?
5. **Performance context** — Is the observed signal persistent rather than temporary?
6. **Technical context** — Could indexing, canonicalization, internal links, or measurement explain the signal?
7. **Business value** — Is the page worth the expected review and refresh effort?
8. **Recent edits** — Was the page already changed recently?

The ranking system should therefore be treated as **ranking plus explanation**, while the final content decision remains human-owned.

---

## Automation boundary

The system should not automatically:

- publish rewritten content;
- delete content;
- alter factual claims;
- modify prices or legally important information;
- modify medical or safety-critical information;
- make irreversible SEO changes;
- declare a page objectively "low quality";
- claim that a refresh caused an improvement;
- make client-specific decisions without owner review.

The appropriate automation boundary is:

> **Model ranking + explanation → Human review → Human decision**

---

## Monitoring the recommendations

The action queue itself can become stale as the underlying data or observed relationships change.

The monitoring plan should track:

- changes in the declining-label distribution;
- changes in feature distributions;
- changes in the distribution of model scores;
- performance on newly labelled, properly held-out data.

A retraining decision should follow investigation rather than happen automatically.

Potential causes of change include data-quality issues, measurement changes, client-mix changes, seasonality, changes in the content population, or changes in the target definition.

The strongest signal for revisiting the model is degradation on newly labelled data evaluated with the same grouped validation logic.

---

## Final recommendation

Based on the measured results, the recommended deployment pattern is to use the Random Forest as a **human-review prioritization system**.

The strongest evidence from this project is that the model concentrated more labeled declining pages near the top of the review queue than the transparent rule baseline on the client-grouped holdout.

The practical recommendation is therefore:

> **Use the model to decide what to inspect first, use reason codes to understand the observed signals, and require human review before making content changes.**

This keeps the system aligned with the evidence produced by the project while avoiding unsupported causal or fully automated claims.
